Task 1: Regex Fundamentals (Character Classes & Quantifiers)

```python
import re

print("Part A")
text = "Natural Language Processing started in 2024. Python 3.12 is useful for NLP! AI, Data, and Models are important."
digits = re.findall(r"\d", text)
print("All digits:", digits)

print("\nPart B")
capital_words = re.findall(r"\b[A-Z][a-zA-Z]*\b", text)
print("Words starting with a capital letter:", capital_words)

print("\nPart C")
four_digits = re.findall(r"\b\d{4}\b", text)
print("Sequences of exactly 4 digits:", four_digits)

print("\nPart D")
words_3_to_6 = re.findall(r"\b[A-Za-z]{3,6}\b", text)
print("Words between 3 to 6 characters:", words_3_to_6)
```

Task 2: Pattern-Based Extraction (Real World Data)

```python
import re

print("Part A")
text = """Contact contact2sufyan@gmail.com or sufyanafzalmalik@gmail.com for more information.
Call 0300-1234567 or 0312-7654321.
Order IDs are ORD-1023 and ORD-2045.
The delivery dates are 15/09/2026 and 28/10/2026.
The discount is 45% and the second offer is 12.5%."""
emails = re.findall(r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b", text)
print("Email addresses:", emails)

print("\nPart B")
phone_numbers = re.findall(r"\b03\d{2}-\d{7}\b", text)
print("Phone numbers:", phone_numbers)

print("\nPart C")
dates = re.findall(r"\b\d{2}/\d{2}/\d{4}\b", text)
print("Dates:", dates)

print("\nPart D")
percentages = re.findall(r"\b\d+(?:\.\d+)?%", text)
print("Percentage values:", percentages)
```

Task 3: Word Tokenization (NLTK vs spaCy)

```python
import nltk
import spacy

print("Part A")
paragraph = """Natural Language Processing is interesting, but it's not always easy.
I don't understand every NLP concept at first.
Python helps students learn these concepts, and practice makes them better.
NLP tools can process words, numbers, and punctuation."""
nltk.download("punkt")
nltk.download("punkt_tab")
nltk_tokens = nltk.word_tokenize(paragraph)
print("NLTK tokens:")
print(nltk_tokens)
print("Total NLTK tokens:", len(nltk_tokens))

print("\nPart B")
nlp = spacy.blank("en")
spacy_doc = nlp(paragraph)
spacy_tokens = [token.text for token in spacy_doc]
print("spaCy tokens:")
print(spacy_tokens)
print("Total spaCy tokens:", len(spacy_tokens))

print("\nPart C")
print("Difference 1: NLTK splits contractions such as \"it's\" into \"it\" and \"'s\".")
print("Difference 2: spaCy keeps contractions such as \"it's\" as one token.")

print("\nPart D")
print("Total NLTK tokens:", len(nltk_tokens))
print("Total spaCy tokens:", len(spacy_tokens))
```

Task 4: Token Attributes (Categorizing Tokens)

```python
import spacy

print("Part A")
sentence = "Sufyan bought 3 books for $50 and contacted contact2sufyan@gmail.com yesterday."
nlp = spacy.blank("en")
doc = nlp(sentence)
print("Sentence:", sentence)
print("Doc:", doc)

print("\nPart B")
for token in doc:
    print("Token:", token.text, "is_alpha:", token.is_alpha)

print("\nPart C")
for token in doc:
    print("Token:", token.text, "like_num:", token.like_num, "like_email:", token.like_email)

print("\nPart D")
currency_tokens = []
for token in doc:
    if token.is_currency:
        currency_tokens.append(token.text)
print("Currency tokens:", currency_tokens)
```
